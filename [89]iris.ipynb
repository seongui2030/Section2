  - `0`: Setosa (세토사)

🔍 EDA(탐색적 데이터 분석) 관점의 해석
1. 데이터의 응집력 (Setosa의 독보적인 일관성)

- setosa 품종은 꽃잎 길이와 너비의 표준편차가 0.1 안팎으로 극도로 작습니다. 이는 세토사 개체들은 서로 매우 흡사한 크기의 꽃잎을 가지고 있으며, 데이터가 한곳에 단단하게 뭉쳐 있다는 것을 의미합니다. 머신러닝 분류 모델 입장에서 가장 명확하게 구별할 수 있는 이유가 바로 이 데이터의 응집력 때문입니다.

2. 특성별 변동성 비교 (Petal vs Sepal)

- 모든 품종에서 공통적으로 꽃잎(petal) 관련 특성의 표준편차(최소 0.10 ~ 최대 0.55)가 꽃받침(sepal) 관련 특성(약 0.31 ~ 0.63)에 비해 전반적으로 작거나 뚜렷한 특징을 보입니다.

- 이는 붓꽃을 분류할 때 꽃받침보다 꽃잎의 크기(길이와 너비)가 품종을 구분하는 훨씬 더 결정적이고 안정적인 지표(특성)가 된다는 점을 데이터의 흩어짐(산포도)을 통해 미리 짐작해 볼 수 있습니다.

- 모델 예측 난이도 시사점

- virginica는 꽃받침과 꽃잎 길이의 표준편차 수치가 가장 큽니다. 즉, 개체 간 편차가 크기 때문에 때로는 versicolor와 경계선상에서 헷갈릴 수 있는 데이터가 존재할 가능성을 이 표준편차 지표를 통해 유추할 수 있습니다.

이처럼 표준편차를 살펴보면 단순히 평균값만 볼 때는 알 수 없었던 "어떤 품종이 균일하고, 어떤 품종이 개체별 차이가 심한지"를 파악할 수 있습니다!

#### 1) 종속변수와 설명변수 구분

### 5. 교차 검증
- K-Fold Cross Validation(K-겹 교차검증) : cross_val_score() 함수

````python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

# 데이타 불러오기
iris = pd.read_csv("./data/iris.csv")

# 데이타 전처리
iris['species'] = iris['species'].str.replace('setosa', '0').replace('versicolor', '1').replace('virginica', '2').astype('int32')

# 데이타 분리하기
X = iris.iloc[:, :-1]
y = iris.iloc[:, -1]

X_train, X_test,y_train, y_test= train_test_split(X, y,
                                                  test_size=0.3,
                                                  random_state=0)

clf = DecisionTreeClassifier( )   # 1) 모델 선택
clf.fit(X_train, y_train)         # 2) 학습 - fit()
y_pred = clf.predict(X_test)      # 3) 예측 - predict()

print('예측 정확도: %.4f' % accuracy_score(y_test, y_pred))
````

````python
from sklearn.model_selection import cross_val_score

score = cross_val_score(clf ,X_train, y_train, cv=10 )
print(score)
print(score.mean())
````

### 5. 교차 검증
- K-Fold Cross Validation(K-겹 교차검증) : cross_val_score() 함수

````python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
import numpy as np

# 데이타 불러오기
iris = pd.read_csv("./data/iris.csv")

# 데이타 전처리
iris['species'] = iris['species'].str.replace('setosa', '0').replace('versicolor', '1').replace('virginica', '2').astype('int32')

# 데이타 분리하기
X = iris.iloc[:, 1:]
y = iris.iloc[:, 1]

X_train, X_test,y_train, y_test= train_test_split(X, y,
                                                  test_size=0.3,
                                                  random_state=0)

lr = LinearRegression()      # 1) 모델 선정
lr.fit(X_train, y_train)     # 2) 모델 학습
y_pred = lr.predict(X_test)  # 3) 예측

print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE :", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R^2 :", r2_score(y_test, y_pred))
````

#### 10-Fold 교차 검증

````python
from sklearn.model_selection import cross_val_score

score = cross_val_score(lr ,X_train, y_train, cv=10 )
print(score)
print(score.mean())
````

## 사이킷런(Scikit-learn)의 의사결정 나무(Decision Tree) 모델이 학습한 결정 규칙과 모델 선택(분류) 과정을 시각적으로 확인
- scikit-learn의 plot_tree 함수를 사용하거나
- export_text를 통해 텍스트 형태로 출력할 수 있습니다.

### 지니 불순도의 의미와 특징
- 최대값 (0.5): 이진 분류 기준, 데이터가 정확히 반반(50%:50%) 섞여 있을 때 불순도가 가장 높습니다. 모델 입장에서 어떤 클래스로 분류해야 할지 알기 가장 혼란스러운 상태입니다.
- 최소값 ($0.0$): 노드 안의 모든 데이터가 단 하나의 클래스로만 이루어져 있을 때(순수 100%) 불순도가 0이 됩니다. 의사결정 나무가 도달하고자 하는 최종 목표 상태입니다.

````python
# 방법 1: plot_tree를 이용해 시각적 트리 그림으로 출력
# 트리의 분기 조건, 지니 불순도(Gini), 샘플 수 등을 이미지 형태로 직접 확인

import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# 1. 시각화 설정
plt.figure(figsize=(16, 10))

# 2. plot_tree 함수 호출
plot_tree(clf,
          feature_names=X.columns,        # 특성 이름 (sepal_length, sepal_width 등)
          class_names=['0', '1', '2'],    # 타겟 클래스 이름 (또는 origin_target)
          filled=True,                    # 노드에 색상 채우기 (순도가 높을수록 진함)
          rounded=True,                   # 테두리를 둥글게 표시
          fontsize=10)

plt.show()
````

````python
from sklearn.tree import export_text

# 특성 이름 지정 후 텍스트 규칙 출력
tree_rules = export_text(clf, feature_names=list(X.columns))
print(tree_rules)
````
