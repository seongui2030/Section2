```python
import os
import urllib.request

# 1. ./data 폴더 생성 (이미 존재하면 무시)
os.makedirs("./data", exist_ok=True)

# 2. 온라인 Iris 데이터셋을 ./data/iris.csv 경로로 다운로드
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
urllib.request.urlretrieve(url, "./data/iris.csv")
```

### 3. 데이타 분리하기
- 1) 종속변수와 설명변수 구분

#### 1) 종속변수와 설명변수 구분

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

#### 2) 학습 데이터셋과 테스트 데이터셋 분리: train_test_split() 함수

```python
from sklearn.model_selection import train_test_split

# 학습데이터셋과 테스트데이터셋 구분 - 70:30
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.3,
                                                    random_state=0)
```

### 4. 지도학습 - 분류

#### 1) 분류 모델

```python
# 지도학습 - 분류 : 의사결정트리
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier()  # 1) 모델 선정
clf.fit(X_train, y_train)       # 2) 모델 학습
y_pred = clf.predict(X_test)    # 3) 예측
clf.score(X_test, y_test)       # 4) 평가(정확도)
```

#### 2) 분류 모델 평가

```python
# 분류 모델의 평가
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score
y_train_pred = clf.predict(X_train) # 학습 데이타 예측
y_test_pred = clf.predict(X_test)   # 테스트 데이타 예측
print('학습데이타 예측 정확도 : %.4f' % accuracy_score(y_train, y_train_pred))
```

### 5. 교차 검증
- K-Fold Cross Validation(K-겹 교차검증) : cross_val_score() 함수

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

# 데이타 불러오기
df = pd.read_csv("./data/iris.csv")

# 데이타 전처리
df['species'] = df['species'].str.replace('setosa', '0').replace('versicolor', '1').replace('virginica', '2').astype('int32')

# 데이타 분리하기
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

X_train, X_test,y_train, y_test= train_test_split(X, y,
                                                  test_size=0.3,
                                                  random_state=0)

clf = DecisionTreeClassifier( )   # 1) 모델 선택
clf.fit(X_train, y_train)         # 2) 학습 - fit()
y_pred = clf.predict(X_test)      # 3) 예측 - predict()

print('예측 정확도: %.4f' % accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import cross_val_score

score = cross_val_score(clf ,X_train, y_train, cv=10 )
print(score)
print(score.mean())
```

### 5. 교차 검증
- K-Fold Cross Validation(K-겹 교차검증) : cross_val_score() 함수

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
import numpy as np

# 데이타 불러오기
df = pd.read_csv("./data/iris.csv")

# 데이타 전처리
df['species'] = df['species'].str.replace('setosa', '0').replace('versicolor', '1').replace('virginica', '2').astype('int32')

# 데이타 분리하기
X = df.iloc[:, 1:]
y = df.iloc[:, 1]

X_train, X_test,y_train, y_test= train_test_split(X, y,
                                                  test_size=0.3,
                                                  random_state=0)

lr = LinearRegression()      # 1) 모델 선정
lr.fit(X_train, y_train)     # 2) 모델 학습
y_pred = lr.predict(X_test)  # 3) 예측

print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE :", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R^2 :", r2_score(y_test, y_pred))
```

#### 10-Fold 교차 검증

```python
from sklearn.model_selection import cross_val_score

score = cross_val_score(lr ,X_train, y_train, cv=10 )
print(score)
print(score.mean())
```
