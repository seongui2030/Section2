```python
import os
import urllib.request

# 1. ./data 폴더 생성 (이미 존재하면 무시)
os.makedirs("./data", exist_ok=True)

# 2. milknew.csv 파일을 ./data/milknew.csv 경로로 다운로드
url = "https://raw.githubusercontent.com/seongui2030/Section2/main/evcharging.csv"
urllib.request.urlretrieve(url, "./data/evcharging.csv")
print("다운로드 완료!")
```

```python
import pandas as pd
df = pd.read_csv('/content/data/evcharging.csv')
df.head(3)
```

[093] 소스 코드 Ⅱ-25 모델 생성하기(K-평균 알고리즘)

```python
from sklearn.cluster import KMeans
KM = KMeans(n_clusters = 5, n_init = 10)
KM.fit(df)
```

[093] 소스 코드 Ⅱ-26 군집화 확인하기

```python
KM.labels_
```

[093] 소스 코드 Ⅱ-27 결과 확인하기
