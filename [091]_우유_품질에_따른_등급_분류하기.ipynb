[091] 소스 코드 Ⅱ-20 데이터 읽기

```python
import os
import urllib.request

# 1. ./data 폴더 생성 (이미 존재하면 무시)
os.makedirs("./data", exist_ok=True)

# 2. milknew.csv 파일을 ./data/milknew.csv 경로로 다운로드
url = "https://raw.githubusercontent.com/seongui2030/Section2/main/milknew.csv"
urllib.request.urlretrieve(url, "./data/milknew.csv")
print("다운로드 완료!")
```

```python
import pandas as pd
df = pd.read_csv('/content/data/milknew.csv')
df.head(3)
```

[091] 소스 코드 Ⅱ-21 데이터 전처리하기(핵심 속성 추출과 데이터 분할)

```python
X = df.drop(columns = ['Grade'])
y = df['Grade']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y)
print('train data 개수:', X_train.shape)
print('test data 개수:', X_test.shape)
```

[091] 소스 코드 Ⅱ-22 모델 생성하기(K-최근접 이웃)

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors = 5)
model.fit(X_train, y_train)
```

[091] 소스 코드 Ⅱ-23 모델의 성능 평가하기

```python
model.score(X_test, y_test)
```
