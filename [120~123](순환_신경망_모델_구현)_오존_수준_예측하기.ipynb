```python
import pandas as pd
oz = pd.read_csv('/content/oz_new.csv')
```

[121] 소스 코드 Ⅱ-57 데이터 전처리하기(핵심 속성 추출과 형태 변형)

```python
oz = oz.loc[:, 'feat_1':]
oz = oz.apply(pd.to_numeric)
oz2 = oz.dropna()
```

[121] 소스 코드 Ⅱ-58 데이터 전처리하기(변수 분할)

```python
X = oz2.loc[:,'feat_1':'feat_72']
y = oz2['feat_73']
```

[121] 소스 코드 Ⅱ-59 데이터 전처리하기(데이터 분할)

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size = 0.3)
y_train, y_test = train_test_split(y, test_size = 0.3)
```

[121] 소스 코드 Ⅱ-60 데이터 확인하기(훈련 데이터 형태)

```python
print(X_train.shape)
print(y_train.shape)
```

[121] 소스 코드 Ⅱ-61 데이터 확인하기(테스트 데이터 형태)

```python
print(X_test.shape)
print(y_test.shape)
```

[122] 소스 코드 Ⅱ-62 데이터 전처리하기(데이터 변형)

```python
X_train = X_train.to_numpy()
X_train = X_train.reshape(-1, 1, 72)
X_test = X_test.to_numpy()
X_test = X_test.reshape(-1, 1, 72)
y_train = y_train.to_numpy()
y_train = y_train.reshape(-1, 1)
y_test = y_test.to_numpy()
y_test = y_test.reshape(-1, 1)
```

[122] 소스 코드 Ⅱ-63 모듈 불러오기

```python
from keras.models import Sequential
from keras.layers import LSTM, Dense
```

[122] 소스 코드 Ⅱ-64 모델 생성하기

```python
model = Sequential()
model.add(LSTM(128, input_shape = (1, 72)))
model.add(Dense(1, activation = 'sigmoid'))
```

[122] 소스 코드 Ⅱ-65 모델 설정하기

```python
model.compile( loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
```

[123] 소스 코드 Ⅱ-66 데이터 학습시키기

```python
model.fit(X_train, y_train, epochs = 10)
```

[123] 소스 코드 Ⅱ-67 모델의 성능 평가하기

```python
model.evaluate(X_test, y_test)
```
